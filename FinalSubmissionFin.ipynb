```python
# !pip install beautifulsoup4

import requests
from bs4 import BeautifulSoup
import re # Import regular expressions

def extract_fund_ids_from_index_page(index_page_url,fund_ids):
    """
    Fetches a Morningstar index page and extracts fund identifiers
    from links matching a specific pattern.

    Args:
        index_page_url (str): The URL of the Morningstar index page
                              (e.g., https://www.morningstar.in/funds/adityabirla.aspx)

    Returns:
        list: A list of extracted fund identifiers.
    """
    print(f"Fetching index page: {index_page_url}")
    # fund_ids = []
    # Regex pattern to match the fund URL structure and capture the fund ID
    # /mutualfunds/ followed by the fund ID (alphanumeric), then / and something else
    # We capture the part after /mutualfunds/ and before the next /
    fund_id_pattern = re.compile(r'/mutualfunds/([^/]+/[^/]+)/overview\.aspx')

    try:
        response = requests.get(index_page_url)
        response.raise_for_status() # Raise an exception for HTTP errors

        soup = BeautifulSoup(response.content, 'html.parser')

        # Find all anchor tags that have an href attribute
        for link in soup.find_all('a', href=True):
            href = link['href']

            # Check if the href matches the fund ID pattern
            match = fund_id_pattern.search(href)
            if match:
                fund_id = match.group(1) # Extract the captured group (the fund ID)
                # Optional: Further validation if needed (e.g., check length, starts with 'f')
                if fund_id.lower().startswith('f') and len(fund_id) > 5: # Basic validation
                    fund_ids.append(fund_id)
                    # print(f"Extracted ID: {fund_id} from {href}") # Uncomment to see extracted IDs

        # Remove duplicates and return the list
        # Using a set to handle duplicates, then convert back to list
        unique_fund_ids = list(set(fund_ids))
        print(f"Found {len(unique_fund_ids)} unique potential fund IDs.")
        return unique_fund_ids

    except requests.exceptions.RequestException as e:
        print(f"Error fetching index page {index_page_url}: {e}")
        return []
    except Exception as e:
        print(f"An unexpected error occurred during processing: {e}")
        return []

# --- Example Usage ---
fund_ids=[]
for i in range(len(fund_house_names_lower)):
    fund_house_name = fund_house_names_lower[i]
    index_page_url = f"https://www.morningstar.in/funds/{fund_house_name}.aspx"
    fund_ids = extract_fund_ids_from_index_page(index_page_url,fund_ids)
# index_page_url_to_process = "https://www.morningstar.in/funds/adityabirla.aspx"

# fund_identifiers = extract_fund_ids_from_index_page(index_page_url_to_process)

if fund_identifiers:
    print("\nList of extracted fund IDs:")
    for fund_id in fund_identifiers:
        print(fund_id)
else:
    print("\nNo fund IDs were extracted.")

# You can now use this list (fund_identifiers) to iterate and fetch data
# using the previously developed fetch_and_process_fund_data function.
# Example:
# from your_module import fetch_and_process_fund_data # Assuming the function is in a module
#
# all_funds_data = {}
# for fund_id in fund_identifiers:
#     fund_data = fetch_and_process_fund_data(fund_id) # Use the function from the previous step
#     if fund_data:
#         all_funds_data[fund_id] = fund_data
#
# print(f"\nProcessed data for {len(all_funds_data)} funds.")
```

```python
def split_identifiers_into_id_and_name(combined_identifiers_list):
    """
    Splits a list of combined fund identifiers (fundid/fund_name)
    into two separate lists: one for fund IDs and one for fund names.

    Args:
        combined_identifiers_list (list): A list of strings in the format "fundid/fund_name".

    Returns:
        tuple: A tuple containing two lists: (list of fund IDs, list of fund names).
               Returns ([], []) if the input list is empty or None.
    """
    fund_ids = []
    fund_names = []

    if not combined_identifiers_list:
        print("Input list is empty or None.")
        return [], []

    for combined_id in combined_identifiers_list:
        # Check if the string contains the expected slash delimiter
        if '/' in combined_id:
            # Split the string at the first occurrence of '/'
            parts = combined_id.split('/', 1) # Use maxsplit=1 to handle potential slashes in the name

            fund_id = parts[0]
            fund_name = parts[1]

            fund_ids.append(fund_id)
            fund_names.append(fund_name)
        else:
            print(f"Warning: Skipping unexpected format in list element: {combined_id}")
            # You might want to handle this case differently,
            # e.g., append None or a default value, or raise an error.


    return fund_ids, fund_names
```

```python
import requests
import json
import time
from bs4 import BeautifulSoup
from urllib.parse import quote
fund_ids = [fund_id.upper() for fund_id in fund_ids]

# --- Assume these lists are available from previous steps ---
# fund_ids = [...] # List of all fund IDs you want to process
# fund_names = [...] # List of fund names corresponding to fund_ids

# Base URL for the Morningstar India website to GET THE TOKEN
MORNINGSTAR_IN_BASE_URL = "https://www.morningstar.in/mutualfunds/"

# Base API URL structure to GET THE DATA
# Let's try increasing the numbers for holdings to see if that helps
# Also, explicitly specifying the component for bond holdings might be needed
# Based on your initial code attempting to access 'boldHoldingPage',
# there might be separate components or sections for equity vs. bond holdings.
# The current component is 'sal-mip-holdings'. Maybe there's another?
# Let's stick with the current component but maximize numbers first.
BASE_API_URL = "https://api-global.morningstar.com/sal-service/v1/fund/portfolio/holding/v2/{fund_id}/data?premiumNum=5000&freeNum=5000&hideesg=true&languageId=en&locale=en&clientId=RSIN_SAL&benchmarkId=mstarorcat&component=sal-mip-holdings&version=4.63.0&access_token={access_token}"


# Dictionary to store the JSON data for each fund
all_funds_json_data = {}

# Optional: Add delays to be polite and avoid being blocked
TOKEN_REQUEST_DELAY_SECONDS = 0 # Increased slightly
API_REQUEST_DELAY_SECONDS = 0 # Increased slightly

# --- Functions from previous steps (re-included for clarity) ---
def build_detailed_portfolio_url(fund_id, fund_name):
    """
    Builds the Detailed Portfolio URL for a fund (where the token is found).
    """
    encoded_fund_name = quote(fund_name)
    url = f"{MORNINGSTAR_IN_BASE_URL}{fund_id}/{encoded_fund_name}/detailed-portfolio.aspx"
    return url

def extract_access_token_from_html(html_content):
    """
    Parses HTML content to find the meta tag containing the access token.
    """
    soup = BeautifulSoup(html_content, 'html.parser')
    access_token_meta = soup.find('meta', attrs={'name': 'accessToken'})
    if access_token_meta and 'content' in access_token_meta.attrs:
        return access_token_meta['content']
    return None

# --- Main Loop to Fetch Data for Each Fund ---

if len(fund_ids) != len(fund_names):
    print("Error: The list of fund IDs and fund names do not match in length. Cannot proceed.")
else:
    for i in range(len(fund_ids)):
        fund_id = fund_ids[i]
        fund_name = fund_names[i][0] if isinstance(fund_names[i], list) else fund_names[i]


        print(f"\n--- Processing Fund: {fund_name} ({fund_id}) ---")

        # --- Step 1: Fetch the detailed portfolio page to get a FRESH token ---
        detailed_portfolio_url = build_detailed_portfolio_url(fund_id, fund_name)
        print(f"Fetching token from: {detailed_portfolio_url}")
        access_token = None

        try:
            token_page_response = requests.get(detailed_portfolio_url)
            token_page_response.raise_for_status()
            html_content = token_page_response.text

            access_token = extract_access_token_from_html(html_content)

            if access_token:
                print(f"Successfully extracted fresh token for {fund_id}.")
            else:
                print(f"Could not extract access token from the page for {fund_id}. Skipping data fetch.")
                all_funds_json_data[fund_id] = {"error": "Could not obtain access token"}
                continue

        except requests.exceptions.RequestException as e:
            print(f"Error fetching token page for {fund_id}: {e}. Skipping data fetch.")
            all_funds_json_data[fund_id] = {"error": f"Error fetching token page: {e}"}
            continue
        except Exception as e:
             print(f"An unexpected error occurred during token extraction for {fund_id}: {e}. Skipping data fetch.")
             all_funds_json_data[fund_id] = {"error": f"Unexpected error during token extraction: {e}"}
             continue

        if TOKEN_REQUEST_DELAY_SECONDS > 0:
             time.sleep(TOKEN_REQUEST_DELAY_SECONDS)


        # --- Step 2: Use the FRESH token to fetch the data from the API ---
        api_url = BASE_API_URL.format(fund_id=fund_id, access_token=access_token)
        print(f"Attempting API call for {fund_id}...") # Added print
        # print(f"API Data URL: {api_url}") # Uncomment to see the full API URL


        try:
            data_response = requests.get(api_url)
            # **Crucially, inspect the response status code and content**
            print(f"API Response Status Code: {data_response.status_code}") # Print status code
            if data_response.status_code != 200:
                 print(f"API returned non-200 status code. Response text:")
                 print(data_response.text[:500]) # Print first 500 chars of response if not 200
                 # Optionally raise for status here if you want it to throw an error
                 # data_response.raise_for_status()

            # **Try to parse JSON, even if status code is not ideal, to see what data is there**
            try:
                data = data_response.json()
                print(f"Successfully parsed JSON response for {fund_id}.") # Added print
                # **Inspect the keys in the returned JSON**
                print(f"Top-level keys in JSON response: {list(data.keys()) if isinstance(data, dict) else data}") # Added print
                all_funds_json_data[fund_id] = data
            except json.JSONDecodeError:
                print(f"Error: Could not decode JSON response for {fund_id}.")
                print(f"Raw response text (first 500 chars): {data_response.text[:500]}") # Print raw text if not JSON
                all_funds_json_data[fund_id] = {"error": "JSON decode error from data API", "raw_response": data_response.text[:500]}


        except requests.exceptions.RequestException as e:
            print(f"Error fetching data for {fund_id}: {e}")
            all_funds_json_data[fund_id] = {"error": f"Error fetching data API: {e}"}
        except Exception as e:
            print(f"An unexpected error occurred for {fund_id}: {e}")
            all_funds_json_data[fund_id] = {"error": f"Unexpected error from data API: {e}"}

        if API_REQUEST_DELAY_SECONDS > 0:
            time.sleep(API_REQUEST_DELAY_SECONDS)


print(f"\n--- Finished Fetching All Data ---")
print(f"Collected data for {len(all_funds_json_data)} funds.")

# --- Format for PDF (as before) ---
formatted_json_output = ""
for fund_id, data in all_funds_json_data.items():
    formatted_json_output += f"--- Data for Fund ID: {fund_id} ---\n"
    try:
        formatted_json_output += json.dumps(data, indent=4)
    except TypeError:
         formatted_json_output += str(data) # Fallback for non-serializable data
    formatted_json_output += "\n\n"

print("\n--- Data Formatting Complete ---")
print("The formatted JSON data is in the 'formatted_json_output' string.")
print("Use the fpdf code from the previous turn, or save 'formatted_json_output' to a file for manual PDF conversion.")

# You would then use the fpdf code from the previous turn with the `formatted_json_output` string
# or save it to a file and convert manually.
```

```python
final_data = {}

for i, fund_id in enumerate(fund_ids, 1):
    try:
        fund_data = all_funds_json_data[fund_id]

        # Initialize holding list
        all_holdings = []

        # Add equity holdings if they exist
        if 'equityHoldingPage' in fund_data and 'holdingList' in fund_data['equityHoldingPage']:
            all_holdings.extend([
                {
                    "security_name": holding['securityName'],
                    "weighting": holding['weighting'],
                    "type": "equity"  # Adding type for clarity
                }
                for holding in fund_data['equityHoldingPage']['holdingList']
            ])

        # Add bond holdings if they exist
        if 'boldHoldingPage' in fund_data and 'holdingList' in fund_data['boldHoldingPage']:
            all_holdings.extend([
                {
                    "security_name": holding['securityName'],
                    "weighting": holding['weighting'],
                    "type": "bond"  # Adding type for clarity
                }
                for holding in fund_data['boldHoldingPage']['holdingList']
            ])

        # Create the fund entry
        final_data[f"fund_{i}"] = {
            "fund_id": fund_data.get('secId', fund_data.get('masterPortfolioId', 'UNKNOWN_ID')),
            "holdingList": all_holdings
        }

    except KeyError as e:
        print(f"Skipping fund {fund_id} - missing key: {e}")
    except Exception as e:
        print(f"Error processing fund {fund_id}: {str(e)}")

# The resulting structure will contain all holdings (equity + bond) for each fund
```
